# Avocado Memo — Supporting Calculations

```python
# Import Packages
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python
# Load the two source files
data_2025 = pd.read_csv("2025Data.csv")
data_2026 = pd.read_csv("2026Data.csv")
```

```python
# Basic check of the source structure
print("2025 shape:", data_2025.shape)
print("2026 shape:", data_2026.shape)
print("Columns:", data_2025.columns.tolist())
```

## 1. 2025 Full-Year Baseline

```python
# Total U.S., all categories, all 52 weeks of 2025
us_2025_full = data_2025[
    (data_2025["Geography"] == "Total U.S.") &
    (data_2025["Period"] <= 52)
].copy()

us_2025_full["Calculated Revenue"] = (
    us_2025_full["Total Bulk and Bags"] *
    us_2025_full["ASP Current Year"]
)

volume_2025_full = us_2025_full["Total Bulk and Bags"].sum()
revenue_2025_full = us_2025_full["Calculated Revenue"].sum()
weighted_asp_2025_full = revenue_2025_full / volume_2025_full

print(f"2025 full-year volume: {volume_2025_full:,.2f}")
print(f"2025 full-year revenue: ${revenue_2025_full:,.2f}")
print(f"2025 full-year weighted ASP: ${weighted_asp_2025_full:.4f}")
```

## 2. 2025 Week 1-32 Baseline For Comparison

```python
us_2025_ytd = data_2025[
    (data_2025["Geography"] == "Total U.S.") &
    (data_2025["Period"] <= 32)
].copy()

us_2025_ytd["Calculated Revenue"] = (
    us_2025_ytd["Total Bulk and Bags"] *
    us_2025_ytd["ASP Current Year"]
)

volume_2025_ytd = us_2025_ytd["Total Bulk and Bags"].sum()
revenue_2025_ytd = us_2025_ytd["Calculated Revenue"].sum()
weighted_asp_2025_ytd = revenue_2025_ytd / volume_2025_ytd

print(f"2025 Weeks 1–32 volume: {volume_2025_ytd:,.2f}")
print(f"2025 Weeks 1–32 revenue: ${revenue_2025_ytd:,.2f}")
print(f"2025 Weeks 1–32 weighted ASP: ${weighted_asp_2025_ytd:.4f}")
```

## 3. Key Market Metrics (Week 1-32)

```python
us_2026_ytd = data_2026[
    (data_2026["Geography"] == "Total U.S.") &
    (data_2026["Period"] <= 32)
].copy()

us_2026_ytd["Calculated Revenue"] = (
    us_2026_ytd["Total Bulk and Bags"] *
    us_2026_ytd["ASP Current Year"]
)

volume_2026_ytd = us_2026_ytd["Total Bulk and Bags"].sum()
revenue_2026_ytd = us_2026_ytd["Calculated Revenue"].sum()
weighted_asp_2026_ytd = revenue_2026_ytd / volume_2026_ytd

print(f"2026 Weeks 1–32 total volume: {volume_2026_ytd:,.2f}")
print(f"2026 Weeks 1–32 total revenue: ${revenue_2026_ytd:,.2f}")
print(f"2026 Weeks 1–32 weighted ASP: ${weighted_asp_2026_ytd:.4f}")
```

## 4. 2026 Category Performance Chart

```python
def summarize_type(df):
    result = (
        df.groupby("Type")
          .apply(
              lambda g: pd.Series({
                  "Volume": g["Total Bulk and Bags"].sum(),
                  "Revenue": (
                      g["Total Bulk and Bags"] *
                      g["ASP Current Year"]
                  ).sum()
              }),
              include_groups=False
          )
    )
    result["Weighted ASP"] = result["Revenue"] / result["Volume"]
    return result

type_2026_ytd = summarize_type(us_2026_ytd)

print("\n2026 Weeks 1–32")
display(type_2026_ytd)
```

## 6. Bulk Size Growth

```python
plu_columns = ["4046", "4225", "4770"]

plu_summary = pd.DataFrame({
    "2025 Weeks 1–32 Volume": us_2025_ytd[plu_columns].sum(),
    "2026 Weeks 1–32 Volume": us_2026_ytd[plu_columns].sum()
})

plu_summary["2026 Volume"] = plu_summary["2026 Weeks 1–32 Volume"]

display(plu_summary)
```

## 7. Bagged Formats

```python
bag_columns = ["TotalBagged", "Sml Bagged", "Lrg Bagged", "X-Lrg Bagged"]

bagged_summary = pd.DataFrame({
    "2025 Weeks 1–32 Volume": us_2025_ytd[bag_columns].sum(),
    "2026 Weeks 1–32 Volume": us_2026_ytd[bag_columns].sum()
})

display(bagged_summary)
```

## 8. Regional Outliers


```python
regions = ["Southeast", "California", "South Central"]

regional_rows = []

for region in regions:
    region_data = data_2026[
        (data_2026["Geography"] == region) &
        (data_2026["Period"] <= 32)
    ].copy()

    region_data["Calculated Revenue"] = (
        region_data["Total Bulk and Bags"] *
        region_data["ASP Current Year"]
    )

    volume = region_data["Total Bulk and Bags"].sum()
    revenue = region_data["Calculated Revenue"].sum()
    weighted_asp = revenue / volume

    regional_rows.append({
        "Region": region,
        "Volume": volume,
        "Revenue": revenue,
        "Weighted ASP": weighted_asp
    })

regional_summary = pd.DataFrame(regional_rows).set_index("Region")

display(regional_summary)
```
